Requirement:

Complete the comparison, then test 5999, 6000, and 6001 rpm. Explain why all three tests are useful.

````python
wheel_speed_rpm = 6000
warning_required = wheel_speed_rpm<=6000 
print(warning_required)

for i in range(5999, 6002):
    warning_required = i <=6000 
    print(warning_required)
````
