### A2. Pass named records through a batch function | 15 minutes

This complete classifier uses the same four rules as Lab 04. It is supplied and correct. Leave it unchanged.

1. CRITICAL if battery is below 15% or voltage is below 22 V.
2. SAFE MODE if battery is below 25% or voltage is below 24 V.
3. WARNING if battery is at most 35%.
4. NOMINAL otherwise.

All measurements here are valid numbers. The new part is looking up the two inputs by their keys. A list holds the records in observation order; each dictionary holds the fields within one record.

````python
def classify_power(battery_percent, voltage_v):
    if battery_percent < 15 or voltage_v < 22:
        return "CRITICAL"
    elif battery_percent < 25 or voltage_v < 24:
        return "SAFE MODE"
    elif battery_percent <= 35:
        return "WARNING"
    else:
        return "NOMINAL"

power_records = [
    {"label": "P1", "battery_percent": 80, "voltage_v": 28},
    {"label": "P2", "battery_percent": 10, "voltage_v": 28},
    {"label": "P3", "battery_percent": 80, "voltage_v": 21},
    {"label": "P4", "battery_percent": 15, "voltage_v": 28},
    {"label": "P5", "battery_percent": 35, "voltage_v": 28},
    {"label": "P1", "battery_percent": 80, "voltage_v": 28},
]
````

The batch function below has one deliberate mistake: it looks up the wrong field in one place. Before running, inspect both arguments passed to `classify_power`.

Repair the mistaken lookup. Keep the rest of the function's behavior: call the classifier once per record, return a new list of named results in the input order, keep repeated labels, and return `[]` for no records. Do not change the input records or print inside the function.

`results.append({"label": ..., "status": ...})` creates a new result dictionary and adds it to the result list. In the display loop, `result` is one of those dictionaries, not the whole list.

````python
def classify_named_batch(records):
    results = []
    for record in records:
        status = classify_power(record["battery_percent"], record["voltage_v"])
        results.append({"label": record["label"], "status": status})
    return results

named_results = classify_named_batch(power_records)
for result in named_results:
    print(result["label"], result["status"])
````

````python
def alert_labels(results):
    labels = []
    for result in results:
        if result["status"] == "CRITICAL" or result["status"] == "SAFE MODE":
            labels.append(result["label"])
    return labels
````

Every input scene is a dictionary with exactly these required fields:

| Key | Type | Meaning |
|---|---|---|
| `scene_id` | string | Observation label; labels may repeat |
| `cloud_percent` | number | Cloud cover percentage |
| `size_mb` | number | Download size in MB |

The fields always exist and have the listed types. There are no missing values, NaN values, or strings in numeric fields. Do not add packages, file input, exception handling, or type checking. `max_cloud` is within 0–100 inclusive; capacity is zero or greater. These are simplified teaching data and rules.

Implement `assess_scene(scene, max_cloud)`. Return exactly one string, using these rules in order:

1. INVALID if cloud cover is outside 0–100 inclusive, or size is zero or negative.
2. CLOUDY if cloud cover exceeds `max_cloud`.
3. ELIGIBLE otherwise, including cloud cover exactly equal to `max_cloud`.

An invalid size wins over the CLOUDY rule. This function knows nothing about the remaining budget. It must use its arguments, leave the scene unchanged, and return without printing.

````python
def assess_scene(scene, max_cloud):
    if scene["cloud_percent"] < 0 or scene["cloud_percent"] > 100 or scene["size_mb"] <= 0:
        return "INVALID"
    elif scene["cloud_percent"] > max_cloud:
        return "CLOUDY"
    else:
        return "ELIGIBLE"
````

### B2. Plan the downloads | about 30 minutes

Implement `plan_downloads(scenes, capacity_mb, max_cloud)`:

1. Start with the supplied capacity. Visit each input record once, in the supplied order. Do not sort or remove records.
2. Call `assess_scene` exactly once per record. Reuse its answer; do not copy its quality rules into the planner.
3. An ELIGIBLE scene becomes DOWNLOAD if its size is no greater than the remaining capacity. Immediately subtract its size when selected.
4. Otherwise an ELIGIBLE scene becomes DEFERRED. Leave capacity unchanged and continue checking later scenes; a smaller scene may still fit.
5. Keep CLOUDY and INVALID statuses unchanged; they use no capacity.
6. Return a new list containing one new result dictionary per input record. Preserve order and repeated labels. Each dictionary has exactly `scene_id`, `status`, `size_mb`, and `remaining_mb`. The last field records the capacity remaining **after** that record's decision. Copy `size_mb` even for an invalid record.

The output statuses are DOWNLOAD, DEFERRED, CLOUDY, or INVALID. ELIGIBLE is an intermediate decision, never a final report status. Empty input returns `[]`. Every call starts fresh, inputs stay unchanged, and the function does not print. A zero capacity is valid.

For example, with capacity 100 MB and threshold 20, a single clear scene of 100 MB returns:

Your plan in words or pseudocode:
Start with an empty results list and remaining capacity equal to capacity_mb
Go through each scene and call assess_scene once
If it is ELIGIBLE compare its size with the remaining capacity

````python
scene_queue = [
    {"scene_id": "A", "cloud_percent": 5, "size_mb": 400},
    {"scene_id": "B", "cloud_percent": 90, "size_mb": 100},
    {"scene_id": "C", "cloud_percent": 10, "size_mb": 700},
    {"scene_id": "D", "cloud_percent": 20, "size_mb": 600},
    {"scene_id": "E", "cloud_percent": -1, "size_mb": 20},
    {"scene_id": "F", "cloud_percent": 90, "size_mb": 0},
    {"scene_id": "G", "cloud_percent": 0, "size_mb": 50},
    {"scene_id": "A", "cloud_percent": 15, "size_mb": 10},
]
````

````python
def plan_downloads(scenes, capacity_mb, max_cloud):
    results = []
    remaining_mb = capacity_mb
    for scene in scenes:
        status = assess_scene(scene, max_cloud)
        if status == "ELIGIBLE":
            if scene["size_mb"] <= remaining_mb:
                status = "DOWNLOAD"
                remaining_mb = remaining_mb - scene["size_mb"]
            else:
                status = "DEFERRED"
        results.append({
            "scene_id": scene["scene_id"],
            "status": status,
            "size_mb": scene["size_mb"],
            "remaining_mb": remaining_mb,
        })
    return results
````

````python
exact_fit = [{"scene_id": "ONE", "cloud_percent": 20, "size_mb": 100}]
expected_fit = [{"scene_id": "ONE", "status": "DOWNLOAD", "size_mb": 100, "remaining_mb": 0}]
assert plan_downloads(exact_fit, 100, 20) == expected_fit
assert plan_downloads([], 100, 20) == []
assert plan_downloads(exact_fit, 0, 20) == [
    {"scene_id": "ONE", "status": "DEFERRED", "size_mb": 100, "remaining_mb": 0}
]
print("Basic planner checks passed; add the full queue check in B5")
````

### B3. Build a report from the decisions | about 15 minutes

Implement `summarize_downloads(results)`. It consumes the planner's returned dictionaries, rather than the original scene queue. Return a dictionary with exactly these five keys:

| Key | Value |
|---|---|
| `DOWNLOAD` | Number of DOWNLOAD records |
| `DEFERRED` | Number of DEFERRED records |
| `CLOUDY` | Number of CLOUDY records |
| `INVALID` | Number of INVALID records |
| `downloaded_mb` | Sum of sizes for DOWNLOAD records only |

Return all five zeros for empty input. Count repeated labels separately. Start fresh on every call; do not print, change inputs, or rerun quality/budget decisions. Input statuses are always one of the four required strings.

A dictionary key can come from a variable: if `status` contains `"CLOUDY"`, `counts[status]` accesses the same entry as `counts["CLOUDY"]`. An update such as `counts[status] = counts[status] + 1` increments that entry. Adapt this pattern rather than hard coding the supplied queue's answers.

Before running, predict the queue's four counts and downloaded MB here: DOWNLOAD 2 and DEFERRED 3 and CLOUDY 1 and INVALID 2 with 1000 MB downloaded.

````python
def summarize_downloads(results):
    counts = {"DOWNLOAD": 0, "DEFERRED": 0, "CLOUDY": 0, "INVALID": 0, "downloaded_mb": 0}
    for result in results:
        status = result["status"]
        counts[status] = counts[status] + 1
        if status == "DOWNLOAD":
            counts["downloaded_mb"] = counts["downloaded_mb"] + result["size_mb"]
    return counts
````
